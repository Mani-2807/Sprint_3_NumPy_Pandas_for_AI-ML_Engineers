## Sample dataset
We use a small fictional customer dataset throughout this notebook. `NaN` represents a missing age.

```python
df = pd.DataFrame({
    'customer': ['Ava', 'Ben', 'Chen', 'Dia', 'Eli', 'Ava'],
    'age': [25, 35, 29, np.nan, 41, 25],
    'city': ['Dallas', 'Austin', 'Dallas', 'Houston', 'Austin', 'Dallas'],
    'spend': [120.0, 250.5, 90.0, 175.0, 310.0, 120.0]
})
print(df)
```

## head() and tail()

### 1. Concept explanation
Show the first or last rows to check the data layout.

### 2. Real-world analogy
Read the first or last few lines of a report.

### 3. Business example
Quickly check the start/end of an import.

### 4. AI/ML example
Spot obvious malformed records before training.

```python
print(df.head(3))
print(df.tail(2))
```

### 5. Output explanation
The first call shows three leading rows; the second shows two trailing rows.

### 6. Best practices
Choose a row count appropriate to the dataset.

### 7. Common mistakes
Printing the whole dataset can overwhelm output.

### 8. Advantages
Fast visual check.

### 9. Limitations
A few rows may not represent all data.

## sample()

### 1. Concept explanation
Returns randomly selected rows, useful for a quick varied inspection.

### 2. Real-world analogy
Open random pages from a large book.

### 3. Business example
Review records from different parts of a sales export.

### 4. AI/ML example
Inspect varied examples in a feature table.

```python
print(df.sample(n=3, random_state=42))
```

### 5. Output explanation
Three sampled rows are displayed; fixed random_state makes this sample reproducible.

### 6. Best practices
Set `random_state` for repeatable demonstrations.

### 7. Common mistakes
Sampling does not guarantee representation of rare groups.

### 8. Advantages
Can reveal patterns missed by the first rows.

### 9. Limitations
Rare issues may not appear.

## info(), describe(), shape, columns, dtypes

### 1. Concept explanation
These summarize structure, descriptive statistics, dimensions, column labels, and storage types.

### 2. Real-world analogy
A report's metadata and statistical summary.

### 3. Business example
Check whether sales imported as numbers and how many records exist.

### 4. AI/ML example
Validate schema and numeric feature ranges.

```python
df.info()
print('shape:', df.shape)
print('columns:', df.columns.tolist())
print('dtypes:\n', df.dtypes)
print(df.describe(include='all'))
```

### 5. Output explanation
The output reports six rows and four columns, types, non-null counts, and summary statistics.

### 6. Best practices
Inspect structure before cleaning or modeling.

### 7. Common mistakes
`describe()` summarizes numeric columns by default unless configured otherwise.

### 8. Advantages
Quick quality overview.

### 9. Limitations
Statistics do not prove data quality or correctness.

## memory_usage()

### 1. Concept explanation
Estimates memory used by each column and optionally the index.

### 2. Real-world analogy
Check how much shelf space each category uses.

### 3. Business example
Find memory-heavy columns in a large report.

### 4. AI/ML example
Optimize training data preparation.

```python
print(df.memory_usage(deep=True))
print('Total bytes:', df.memory_usage(deep=True).sum())
```

### 5. Output explanation
Memory estimates are shown per column and as a total.

### 6. Best practices
Use `deep=True` for a more complete estimate of object data.

### 7. Common mistakes
Memory numbers can vary by Pandas version and object content.

### 8. Advantages
Helps identify optimization opportunities.

### 9. Limitations
Does not measure every temporary allocation.

## nunique() and value_counts()

### 1. Concept explanation
`nunique` counts distinct values; `value_counts` counts frequency of each value.

### 2. Real-world analogy
Count unique types and how often each appears.

### 3. Business example
Find common cities or category imbalance.

### 4. AI/ML example
Inspect target class imbalance and categorical features.

```python
print(df.nunique())
print(df['city'].value_counts(dropna=False))
```

### 5. Output explanation
Distinct counts appear for each column; city frequencies show how often each city occurs.

### 6. Best practices
Use `dropna=False` when missing values should be counted explicitly.

### 7. Common mistakes
High cardinality is not necessarily an error.

### 8. Advantages
Useful for categorical quality checks.

### 9. Limitations
Counts alone do not explain why a category is common.
