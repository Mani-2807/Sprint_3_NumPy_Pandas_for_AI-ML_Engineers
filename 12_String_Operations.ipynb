## Sample text data

```python
df = pd.DataFrame({
    'customer': ['  AVA Smith ', 'Ben JONES', 'chen lee'],
    'email': ['ava@example.com', 'BEN@EXAMPLE.COM', 'invalid-email'],
    'code': ['ORD-1001', 'ORD-1002', 'INV-1003']
})
print(df)
```

## lower(), upper(), and strip()

### 1. Concept explanation
These normalize letter case and remove leading/trailing whitespace.

### 2. Real-world analogy
Clean names written with inconsistent capitalization and extra spaces.

### 3. Business example
Standardize customer names and email addresses.

### 4. AI/ML example
Reduce inconsistent text values before encoding categories.

```python
df['customer_clean'] = df['customer'].str.strip().str.title()
df['email_clean'] = df['email'].str.strip().str.lower()
print(df[['customer_clean', 'email_clean']])
```

### 5. Output explanation
Names are trimmed and title-cased; email addresses are trimmed and lowercased.

### 6. Best practices
Preserve original columns when auditing transformations is important.

### 7. Common mistakes
Case conversion is not a substitute for validating identity.

### 8. Advantages
Simple, readable text cleaning.

### 9. Limitations
Some languages and names have special capitalization rules.

## split() and contains()

### 1. Concept explanation
`split` separates text; `contains` checks whether a pattern appears.

### 2. Real-world analogy
Split a full address into parts or find messages containing a word.

### 3. Business example
Extract email domains or flag a category of notes.

### 4. AI/ML example
Create text-derived features.

```python
print(df['customer'].str.strip().str.split())
print(df['email'].str.contains('example', case=False, na=False))
```

### 5. Output explanation
The first output contains lists of name parts; the second returns True/False values.

### 6. Best practices
Use `na=False` when missing values should count as non-matches.

### 7. Common mistakes
Default contains behavior treats the pattern as regex.

### 8. Advantages
Convenient column-wise text processing.

### 9. Limitations
Unescaped regex characters can change matching behavior.

## replace() and regular expressions

### 1. Concept explanation
String replacement changes matching text; regex describes patterns such as digits, letters, and repeated structures.

### 2. Real-world analogy
Find and replace a pattern across many forms.

### 3. Business example
Standardize order IDs or remove unwanted punctuation.

### 4. AI/ML example
Extract structured signals from logs or text fields.

```python
df['id_number'] = df['code'].str.extract(r'(\d+)')
df['code_prefix'] = df['code'].str.replace(r'-\d+', '', regex=True)
print(df[['code', 'id_number', 'code_prefix']])
```

### 5. Output explanation
The digits are extracted into id_number and numeric suffixes are removed from code_prefix.

### 6. Best practices
Test regex patterns on examples and edge cases.

### 7. Common mistakes
Regex patterns can match too much or too little.

### 8. Advantages
Handles variable text patterns efficiently.

### 9. Limitations
Complex patterns can be hard to maintain.
