## Sample dataset

```python
df = pd.DataFrame({
    'name': ['Ava', 'Ben', 'Chen', 'Dia', 'Eli'],
    'age': [25, 35, 29, 42, 31],
    'city': ['Dallas', 'Austin', 'Dallas', 'Houston', 'Austin'],
    'spend': [120, 250, 90, 175, 310]
})
print(df)
```

## loc and iloc

### 1. Concept explanation
`loc` selects by row/column labels; `iloc` selects by integer positions.

### 2. Real-world analogy
Find a person by name versus by their position in a list.

### 3. Business example
Retrieve a customer's fields or a report's first rows.

### 4. AI/ML example
Select known feature columns and row ranges.

```python
print(df.loc[0, 'name'])
print(df.loc[1:3, ['name', 'spend']])
print(df.iloc[0:2, 0:2])
```

### 5. Output explanation
`loc` uses labels and label slices include the endpoint; `iloc` uses integer positions and stop endpoints are excluded.

### 6. Best practices
Prefer explicit column names for clarity.

### 7. Common mistakes
Confusing label and position indexing can return different rows.

### 8. Advantages
Flexible selection.

### 9. Limitations
Results depend on index labels.

## Boolean indexing and filtering rows

### 1. Concept explanation
Boolean indexing keeps rows where a condition is true.

### 2. Real-world analogy
Choose only people meeting a set of criteria.

### 3. Business example
Find orders above a sales threshold.

### 4. AI/ML example
Filter records for analysis or model cohorts.

```python
high_spend = df[df['spend'] >= 175]
combined = df[(df['age'] >= 30) & (df['city'] == 'Austin')]
print(high_spend)
print(combined)
```

### 5. Output explanation
The first table includes spend values at least 175; the second includes Austin customers age 30 or older.

### 6. Best practices
Wrap each condition in parentheses and use `&` or `|`.

### 7. Common mistakes
Using Python `and` instead of `&` on Series raises an error.

### 8. Advantages
Expressive filtering.

### 9. Limitations
Null values may make comparisons false or unknown.

## query(), isin(), between()

### 1. Concept explanation
`query` expresses conditions as a string; `isin` checks membership; `between` checks a range.

### 2. Real-world analogy
Find people in a guest list or between two ages.

### 3. Business example
Select a chosen set of stores or an order range.

### 4. AI/ML example
Create filtered training cohorts.

```python
print(df.query('spend >= 150 and age < 40'))
print(df[df['city'].isin(['Dallas', 'Houston'])])
print(df[df['age'].between(25, 35)])
```

### 5. Output explanation
Each output contains rows matching its condition.

### 6. Best practices
Use `isin` for membership and `between` for inclusive ranges by default.

### 7. Common mistakes
Be careful with column names that contain spaces in `query`.

### 8. Advantages
Concise readable filters.

### 9. Limitations
Complex expressions can be less clear in strings.

## Sorting data

### 1. Concept explanation
`sort_values` sorts by one or more columns; `sort_index` sorts labels.

### 2. Real-world analogy
Arrange a leaderboard from highest score to lowest.

### 3. Business example
Rank stores by revenue.

### 4. AI/ML example
Inspect high-risk or high-value samples.

```python
print(df.sort_values(['spend', 'age'], ascending=[False, True]))
```

### 5. Output explanation
Rows are sorted by spend descending and age ascending for ties.

### 6. Best practices
Specify sort direction explicitly when important.

### 7. Common mistakes
Sorting returns a new object by default unless configured otherwise.

### 8. Advantages
Makes ranking and review easy.

### 9. Limitations
Sorting does not explain the cause of ranking.
