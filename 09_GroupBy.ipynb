## Sample dataset

```python
sales = pd.DataFrame({
    'region': ['East', 'East', 'West', 'West', 'East', 'West'],
    'product': ['Tea', 'Coffee', 'Tea', 'Coffee', 'Tea', 'Tea'],
    'units': [10, 5, 8, 7, 12, 4],
    'revenue': [100, 150, 80, 210, 120, 40]
})
print(sales)
```

## groupby() and agg()

### 1. Concept explanation
`groupby` divides rows into groups; `agg` calculates one or more summaries per group.

### 2. Real-world analogy
Total up each team's expenses separately.

### 3. Business example
Summarize revenue and units by region or product.

### 4. AI/ML example
Analyze performance by segment or cohort.

```python
summary = sales.groupby('region').agg(total_revenue=('revenue', 'sum'), avg_units=('units', 'mean'), orders=('product', 'count'))
print(summary)
```

### 5. Output explanation
Each region receives total revenue, average units, and row count.

### 6. Best practices
Name aggregation outputs and specify grouping keys clearly.

### 7. Common mistakes
Grouping by the wrong key can hide important differences.

### 8. Advantages
Compact summaries across many groups.

### 9. Limitations
Aggregates lose row-level detail.

## transform()

### 1. Concept explanation
`transform` returns results aligned to the original rows, often group statistics repeated per row.

### 2. Real-world analogy
Write each person's team average beside their score.

### 3. Business example
Compare each order with its region's average revenue.

### 4. AI/ML example
Create group-relative features without changing row count.

```python
sales['region_avg_revenue'] = sales.groupby('region')['revenue'].transform('mean')
print(sales[['region', 'revenue', 'region_avg_revenue']])
```

### 5. Output explanation
Every row retains its original position and gets its region's mean revenue.

### 6. Best practices
Use transform when a group-level value must be attached to each record.

### 7. Common mistakes
Do not expect transform to reduce the number of rows.

### 8. Advantages
Useful for feature engineering.

### 9. Limitations
Some transformations can be memory intensive.

## filter()

### 1. Concept explanation
Group filtering keeps or drops entire groups based on a group-level condition.

### 2. Real-world analogy
Keep teams that have at least five members.

### 3. Business example
Retain regions with enough orders for reliable analysis.

### 4. AI/ML example
Exclude tiny cohorts from some descriptive comparisons, with documented reasoning.

```python
large_regions = sales.groupby('region').filter(lambda group: len(group) >= 3)
print(large_regions)
```

### 5. Output explanation
Only rows from groups with at least three records remain.

### 6. Best practices
State the group-level rule explicitly.

### 7. Common mistakes
Filtering groups based on outcome information can bias model evaluation.

### 8. Advantages
Convenient for cohort filtering.

### 9. Limitations
Removing small groups may create selection bias.

## Pivot tables and crosstab

### 1. Concept explanation
Pivot tables summarize values across row/column categories; crosstab counts combinations of categories by default.

### 2. Real-world analogy
A matrix that compares teams and product types.

### 3. Business example
Compare revenue by region and product, or count orders by status.

### 4. AI/ML example
Explore relationships between categorical features and labels.

```python
pivot = pd.pivot_table(sales, values='revenue', index='region', columns='product', aggfunc='sum', fill_value=0)
print(pivot)
print(pd.crosstab(sales['region'], sales['product']))
```

### 5. Output explanation
The pivot reports revenue per region/product combination; crosstab reports row counts.

### 6. Best practices
Choose the aggregation and missing-combination behavior deliberately.

### 7. Common mistakes
A crosstab count is not a revenue total.

### 8. Advantages
Useful compact comparisons.

### 9. Limitations
Large numbers of categories can create wide sparse tables.
