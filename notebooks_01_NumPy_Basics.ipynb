# 1. What is NumPy?

## Definition 

NumPy stands for **Numerical Python**. It is a Python library used to perform numerical calculations efficiently.

NumPy provides a special data structure called an **array**, which allows us to store and process numerical data easily.

Python lists can store different types of values, but NumPy arrays are designed primarily for numerical operations and can perform calculations on entire arrays efficiently.

## Formula

For an array containing numerical values, basic operations can be performed element by element.

For example:

`result = array + number`

## Simple Numerical Example

Suppose we have the following marks:

- Mathematics: 80
- Science: 90
- English: 85

Adding 5 bonus marks to every subject gives:

`[80, 90, 85] + 5 = [85, 95, 90]`

NumPy allows us to perform this operation on all elements at once.

## Business Example

A company stores the monthly sales of three products as `[1000, 1500, 2000]`. If each product's sales increase by 10%, NumPy can calculate the updated sales values efficiently.

## AI/ML Example

In Machine Learning, datasets contain numerical features such as age, salary, temperature, and examination marks. NumPy arrays help process these numerical values before they are used to train a Machine Learning model.

```python
import numpy as np

marks = np.array([80, 90, 85])

updated_marks = marks + 5

print("Original Marks:", marks)
print("Updated Marks:", updated_marks)
```

## Code Implementation Explanation

- `import numpy as np` imports the NumPy library and gives it the short name `np`.
- `np.array([80, 90, 85])` creates a NumPy array containing three marks.
- `marks + 5` adds 5 to every element without requiring a loop.
- `print()` displays the original and updated marks.


## Interpretation

NumPy performs the addition on every array element automatically. This makes numerical calculations simpler and more convenient for data analysis and AI/ML tasks.

# 2. Why NumPy?

## Definition 

NumPy is used because it makes numerical calculations easier, faster, and more convenient. It supports vectorized operations, multidimensional arrays, mathematical functions, and efficient storage of numerical data.

Unlike a normal Python list, a NumPy array allows us to apply many operations to all its elements directly.

## Formula

For increasing each value by a fixed amount:

\[
y_i = x_i + c
\]

Where:
- \(x_i\) is an original value.
- \(c\) is the added amount.
- \(y_i\) is the updated value.

## Simple Numerical Example

Original values: `[10, 20, 30]`

Add 5 to each value: `[15, 25, 35]`

## Business Example

An online store increases the price of every product by ₹100. NumPy can update all product prices with one operation.

## AI/ML Example

NumPy is useful for preparing datasets, scaling numerical features, calculating errors, and performing mathematical operations used in Machine Learning algorithms.

```python
import numpy as np

prices = np.array([500, 800, 1200])

updated_prices = prices + 100

print("Original Prices:", prices)
print("Updated Prices:", updated_prices)
```

## Code Implementation Explanation

- `prices` stores the product prices in a NumPy array.
- `prices + 100` adds ₹100 to every price using vectorized addition.
- Both arrays are displayed using `print()`.


## Interpretation

NumPy lets us perform the same calculation on multiple values without writing a separate operation for every product. This is useful when working with large datasets.

# 3. NumPy Arrays (ndarray)

## Definition 

A NumPy array is a collection of elements stored in an organized structure. The main NumPy array object is called `ndarray`, meaning *N-dimensional array*.

An array can have one dimension, two dimensions, or more. NumPy arrays generally store elements of a common data type.

## Formula

For a one-dimensional array containing \(n\) elements:

\[
A = [a_1, a_2, \ldots, a_n]
\]

## Simple Numerical Example

`[10, 20, 30, 40]` is a one-dimensional array containing four elements.

A two-dimensional array can be represented as:

\[
A = \begin{bmatrix} 10 & 20 \\ 30 & 40 \end{bmatrix}
\]

## Business Example

A company stores the daily sales of four products in a one-dimensional array. It can store the sales of multiple products across multiple days in a two-dimensional array.

## AI/ML Example

A dataset may represent each row as one observation and each column as a feature. NumPy arrays can represent the numerical feature matrix supplied to a Machine Learning model.

```python
import numpy as np

one_dimensional = np.array([10, 20, 30, 40])

two_dimensional = np.array([
    [10, 20],
    [30, 40]
])

print("1D Array:")
print(one_dimensional)

print("2D Array:")
print(two_dimensional)

print("Array Type:", type(one_dimensional))
```

## Code Implementation Explanation

- `np.array()` converts the supplied sequences into NumPy arrays.
- `one_dimensional` contains four elements in a single dimension.
- `two_dimensional` contains two rows and two columns.
- `type()` identifies the Python object type.

## Interpretation

Both variables are NumPy arrays. Their structures differ because one has a single dimension and the other has two dimensions.

# 4. Creating NumPy Arrays

## Definition 

NumPy provides several functions to create arrays depending on the type of data needed.

Common methods include:

- `np.array()` — creates an array from a list or another sequence.
- `np.zeros()` — creates an array filled with zeros.
- `np.ones()` — creates an array filled with ones.
- `np.arange()` — creates values in a specified range using a step.
- `np.linspace()` — creates a specified number of evenly spaced values.
- `np.full()` — creates an array filled with a chosen value.

## Formula

For `np.arange(start, stop, step)`, values begin at `start` and increase by `step`, stopping before `stop`.

\[
a_k = start + k \times step
\]

## Simple Numerical Example

`np.arange(1, 6)` produces `[1, 2, 3, 4, 5]`.

`np.zeros(3)` produces `[0., 0., 0.]`.

## Business Example

A business can create an initial sales target array filled with zeros and update it as actual sales data becomes available.

## AI/ML Example

Arrays of zeros and ones are useful for initializing numerical data, creating masks, and setting up test inputs for Machine Learning calculations.

```python
import numpy as np

array_from_list = np.array([10, 20, 30])
zeros_array = np.zeros(3)
ones_array = np.ones(3)
range_array = np.arange(1, 6, 1)
evenly_spaced = np.linspace(0, 10, 5)
full_array = np.full(3, 7)

print("Array:", array_from_list)
print("Zeros:", zeros_array)
print("Ones:", ones_array)
print("Range:", range_array)
print("Evenly Spaced:", evenly_spaced)
print("Full:", full_array)
```

## Code Implementation Explanation

- `np.array()` creates an array from existing values.
- `np.zeros(3)` and `np.ones(3)` create three-element arrays.
- `np.arange(1, 6, 1)` generates integers from 1 through 5.
- `np.linspace(0, 10, 5)` generates five evenly spaced values between 0 and 10, including both endpoints.
- `np.full(3, 7)` fills three positions with the value 7.

## Interpretation

Different array creation functions suit different requirements. Choose the function based on whether you need existing values, a numerical range, evenly spaced values, or an array initialized with a constant.

# 5. Dimensions of NumPy Arrays

## Definition 

The dimension of an array describes the number of axes along which its elements are organized.

- **0D array:** A single value, also called a scalar array.
- **1D array:** A sequence of values.
- **2D array:** Rows and columns, like a table.
- **3D array:** A collection of two-dimensional arrays.

NumPy's `ndim` attribute tells us the number of dimensions.

## Formula

For a two-dimensional array, its axes represent rows and columns. In general, the number of dimensions is available as:

\[
\text{Dimensions} = A.ndim
\]

## Simple Numerical Example

- `np.array(5)` has 0 dimensions.
- `np.array([1, 2, 3])` has 1 dimension.
- `np.array([[1, 2], [3, 4]])` has 2 dimensions.

## Business Example

A business might represent one sales value as a 0D array, daily sales as a 1D array, and sales by product and day as a 2D array.

## AI/ML Example

Image data is often represented using two dimensions for grayscale images and three dimensions for color channels, excluding any optional batch dimension.

```python
import numpy as np

array_0d = np.array(5)
array_1d = np.array([1, 2, 3])
array_2d = np.array([[1, 2], [3, 4]])
array_3d = np.array([[[1, 2], [3, 4]]])

print("0D Dimensions:", array_0d.ndim)
print("1D Dimensions:", array_1d.ndim)
print("2D Dimensions:", array_2d.ndim)
print("3D Dimensions:", array_3d.ndim)
```

## Code Implementation Explanation

- Four arrays are created with different structures.
- The `.ndim` attribute returns the number of dimensions for each array.


## Interpretation

The number of dimensions depends on how the data is nested and organized, not on the number of elements it contains.

# 6. Shape of NumPy Arrays

## Definition 

The shape of an array describes the number of elements along each dimension. It is returned as a tuple using the `.shape` attribute.

For a two-dimensional array, the shape is `(rows, columns)`.

## Formula

\[
\text{Shape} = (d_1, d_2, \ldots, d_n)
\]

Here, each \(d_i\) represents the size of the corresponding dimension.

## Simple Numerical Example

An array with two rows and three columns has the shape `(2, 3)`.

## Business Example

A sales table containing five products and three measurements per product has the shape `(5, 3)`.

## AI/ML Example

Machine Learning datasets commonly have the shape `(number of samples, number of features)`. Checking the shape helps ensure that data has the expected structure before training a model.

```python
import numpy as np

sales = np.array([
    [100, 200, 300],
    [150, 250, 350]
])

print("Sales Array:")
print(sales)
print("Shape:", sales.shape)
```

## Code Implementation Explanation

- `sales` contains two rows and three columns.
- `.shape` returns a tuple representing the size of each dimension.


## Interpretation

The shape `(2, 3)` means the array contains two rows and three columns, for a total of six elements.

# 7. Size of NumPy Arrays

## Definition

The size of an array is the total number of elements it contains. The `.size` attribute returns this count regardless of the number of dimensions.

## Formula

\[
\text{Size} = d_1 \times d_2 \times \cdots \times d_n
\]

For a two-dimensional array, this becomes:

\[
\text{Size} = \text{Rows} \times \text{Columns}
\]

## Simple Numerical Example

An array with shape `(2, 3)` has six elements because \(2 \times 3 = 6\).

## Business Example

A company can use the array size to count how many individual sales measurements are stored in a dataset.

## AI/ML Example

Array size helps determine how many numerical values are stored in feature matrices and image data.

```python
import numpy as np

data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Shape:", data.shape)
print("Total Elements:", data.size)
```

## Code Implementation Explanation

- `.shape` returns the dimensions of the array.
- `.size` returns the total number of elements.

## Interpretation

Shape describes how the elements are arranged, while size tells us how many elements exist in total.

# 8. Data Types in NumPy

## Definition 

A data type specifies the kind of value an array stores, such as integers, decimal numbers, or Boolean values.

The `.dtype` attribute identifies the array's data type. NumPy arrays usually use one common data type for their elements, which supports efficient numerical processing.

Common data types include:
- `int32` or `int64` — integers.
- `float32` or `float64` — decimal numbers.
- `bool` — True or False values.

The exact default type can depend on the platform and how the array is created.

## Formula

There is no mathematical formula. The relevant property is:

`array.dtype`

## Simple Numerical Example

An array containing `[10, 20, 30]` can use an integer data type, while `[10.5, 20.5, 30.5]` can use a floating-point data type.

## Business Example

Sales quantities may be stored as integers, while product prices and revenue calculations often require floating-point values.

## AI/ML Example

Machine Learning models use numerical arrays with suitable data types. Using `float32` instead of `float64` can reduce memory usage in some workloads, although precision and model requirements must be considered.

```python
import numpy as np

integer_array = np.array([10, 20, 30])
float_array = np.array([10.5, 20.5, 30.5])
boolean_array = np.array([True, False, True])

print("Integer Data Type:", integer_array.dtype)
print("Float Data Type:", float_array.dtype)
print("Boolean Data Type:", boolean_array.dtype)

float32_array = np.array([1.5, 2.5, 3.5], dtype=np.float32)
print("Specified Data Type:", float32_array.dtype)
```

## Code Implementation Explanation

- Each array is created with values of a particular kind.
- `.dtype` displays the data type selected by NumPy.
- `dtype=np.float32` explicitly requests 32-bit floating-point values.


## Interpretation

Selecting a suitable data type helps control memory use and numerical precision. Avoid changing data types without considering the accuracy your calculations require.

# 9. Indexing in NumPy

## Definition 

Indexing is the process of accessing a specific element in an array using its position.

NumPy uses **zero-based indexing**, meaning the first element has index `0`, the second has index `1`, and so on.

For two-dimensional arrays, we use `[row_index, column_index]`.

## Formula

For a one-dimensional array:

\[
\text{Element} = A[i]
\]

For a two-dimensional array:

\[
\text{Element} = A[i,j]
\]

## Simple Numerical Example

For the array `[10, 20, 30, 40]`, the element at index `2` is `30`.

## Business Example

A business can access the sales value for a particular product stored at a known position in an array.

## AI/ML Example

Indexing helps access a particular feature, observation, or numerical value in a dataset while preparing model inputs.

```python
import numpy as np

marks = np.array([75, 80, 90, 85])

print("First Mark:", marks[0])
print("Third Mark:", marks[2])
print("Last Mark:", marks[-1])

scores = np.array([
    [80, 85, 90],
    [70, 75, 95]
])

print("Row 2, Column 3:", scores[1, 2])
```

## Code Implementation Explanation

- `marks[0]` accesses the first element.
- `marks[2]` accesses the third element.
- `marks[-1]` accesses the last element using negative indexing.
- `scores[1, 2]` accesses the second row and third column.

## Interpretation

Indexing allows us to retrieve individual elements without printing or processing the entire array.

# 10. Slicing in NumPy

## Definition 

Slicing extracts a portion of an array. Instead of selecting one element, we can select a range of elements using the syntax:

`array[start:stop:step]`

The start index is included, but the stop index is excluded. If omitted, the start, stop, and step use their default values.

## Formula

For a slice with start index \(s\), stop index \(e\), and step \(k\), the selected indices are:

\[
s,\ s+k,\ s+2k,\ \ldots < e
\]

## Simple Numerical Example

For `[10, 20, 30, 40, 50]`, the slice `[1:4]` returns `[20, 30, 40]`.

## Business Example

A business can extract sales data for a selected range of days from a larger array.

## AI/ML Example

Slicing can select specific rows or columns from numerical datasets, such as choosing a subset of observations or model features.

```python
import numpy as np

sales = np.array([100, 200, 300, 400, 500])

print("First Three:", sales[:3])
print("Middle Values:", sales[1:4])
print("Every Second Value:", sales[::2])
print("Last Two:", sales[-2:])

data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("First Two Rows:")
print(data[:2, :])
```

## Code Implementation Explanation

- `sales[:3]` selects elements from the beginning up to, but not including, index 3.
- `sales[1:4]` selects indices 1, 2, and 3.
- `sales[::2]` selects every second element.
- `sales[-2:]` selects the final two elements.
- `data[:2, :]` selects the first two rows and all columns.


## Interpretation

Slicing is useful for selecting relevant portions of large arrays. Basic NumPy slicing generally returns a view of the original array rather than a separate copy, so modifying a sliced view can affect the original data.

# 11. Reshaping NumPy Arrays

## Definition 

Reshaping changes the dimensions of an array without changing the order or total number of its elements.

The `reshape()` method is used to convert an array into a different shape.

The total number of elements must remain the same.

## Formula

\[
\text{Original Size} = \text{New Size}
\]

For example:

\[
6 = 2 \times 3
\]

An array of six elements can therefore be reshaped into two rows and three columns.

## Simple Numerical Example

`[1, 2, 3, 4, 5, 6]` can be reshaped into:

`[[1, 2, 3], [4, 5, 6]]`

## Business Example

A company can arrange six daily sales values into two rows and three columns to represent two products across three days.

## AI/ML Example

Reshaping helps convert data into the dimensions required by numerical algorithms or Machine Learning models. The required shape depends on the model and the meaning of the data.

```python
import numpy as np

numbers = np.array([1, 2, 3, 4, 5, 6])

reshaped = numbers.reshape(2, 3)

print("Original Array:")
print(numbers)

print("Reshaped Array:")
print(reshaped)

print("New Shape:", reshaped.shape)
```

## Code Implementation Explanation

- `numbers` contains six elements.
- `reshape(2, 3)` organizes the six elements into two rows and three columns.
- `.shape` confirms the new dimensions.

## Interpretation

Reshaping changes the organization of the data, not the number or order of the values. A reshape may return a view when possible or create a copy when necessary.

# 12. Flattening NumPy Arrays

## Definition 

Flattening converts a multidimensional array into a one-dimensional array.

NumPy provides the `flatten()` method, which returns a new copy of the array's data.

## Formula

For a two-dimensional array:

\[
A =
\begin{bmatrix}
1 & 2 \\
3 & 4
\end{bmatrix}
\]

Flattening in the default row-major order gives:

\[
A_{\text{flat}} = [1, 2, 3, 4]
\]

## Simple Numerical Example

`[[10, 20], [30, 40]]` becomes `[10, 20, 30, 40]`.

## Business Example

A company can flatten a table of numerical measurements into one sequence when a downstream operation requires a one-dimensional representation.

## AI/ML Example

Flattening is often used in neural networks to convert multidimensional image features into a vector before passing them to a dense layer. Other models may require different input shapes.

```python
import numpy as np

data = np.array([
    [10, 20],
    [30, 40]
])

flat_data = data.flatten()

print("Original Array:")
print(data)

print("Flattened Array:")
print(flat_data)
```

## Code Implementation Explanation

- `data` is a two-dimensional array.
- `flatten()` returns a one-dimensional copy containing the elements in row-major order by default.
- `flat_data` contains all four values in a single sequence.


## Interpretation

Flattening is useful when an operation requires one-dimensional input. Because `flatten()` returns a copy, changing the flattened array does not change the original array's elements.

# 13. Copy vs. View in NumPy

## Definition

A **copy** creates a separate array with its own data. Changes to the copied array do not change the original array.

A **view** creates a new array object that shares data with the original array. Changes to shared elements can affect both arrays.

- `copy()` explicitly creates a copy.
- `view()` creates a view that shares the underlying data when applicable.
- Basic slicing commonly creates a view.

## Formula

There is no mathematical formula. The key difference is whether the underlying data is shared.

## Simple Numerical Example

Original array: `[10, 20, 30]`

If a copy's first element is changed to `99`, the original remains `[10, 20, 30]`.

If a view's first element is changed to `99`, the original also reflects that change.

## Business Example

A business analyst might create a copy before modifying source sales data to preserve the original values for comparison.

## AI/ML Example

Understanding copies and views helps prevent unintended changes when preparing datasets, selecting features, or transforming numerical arrays. Shared data can save memory, but it requires careful handling.

```python
import numpy as np

original = np.array([10, 20, 30])

copied = original.copy()
viewed = original.view()

copied[0] = 99

print("After Modifying Copy:")
print("Original:", original)
print("Copy:", copied)

viewed[1] = 88

print("\nAfter Modifying View:")
print("Original:", original)
print("View:", viewed)
```

## Code Implementation Explanation

- `original.copy()` creates a separate copy.
- `original.view()` creates another array object that shares the original data.
- Changing `copied[0]` does not change `original[0]`.
- Changing `viewed[1]` changes the shared data and is visible in `original`.


## Interpretation

Use a copy when you need independent data. Use a view when sharing the original data is appropriate and you understand the effects of modifying it.

# 14. Broadcasting in NumPy

## Definition 

Broadcasting is a NumPy feature that allows arrays with compatible shapes to participate in arithmetic operations without manually repeating their values.

For example, NumPy can add a single number to every element of an array. It can also perform operations between arrays of different shapes when their dimensions are compatible.

Two dimensions are compatible when they are equal or when one of them is `1`. NumPy compares dimensions starting from the right.

## Formula

For an array \(A\) and a scalar \(c\):

\[
B_{ij} = A_{ij} + c
\]

## Simple Numerical Example

Suppose:

`A = [[10, 20], [30, 40]]`

Adding `5` produces:

`[[15, 25], [35, 45]]`

The value `5` is applied to every element.

## Business Example

A store adds ₹50 to the price of every product. Broadcasting allows the price adjustment to be applied to an entire array at once.

## AI/ML Example

Broadcasting is used when normalizing data, subtracting feature means, adding model biases, and performing numerical operations across batches of data.

```python
import numpy as np

prices = np.array([
    [100, 200, 300],
    [400, 500, 600]
])

updated_prices = prices + 50

print("Original Prices:")
print(prices)

print("Updated Prices:")
print(updated_prices)

# Broadcasting a row across multiple rows
base_prices = np.array([10, 20, 30])
adjustments = prices + base_prices

print("Row Broadcasting:")
print(adjustments)
```

## Code Implementation Explanation

- `prices + 50` adds the scalar value 50 to every element.
- `base_prices` contains three values and has shape `(3,)`.
- `prices` has shape `(2, 3)`.
- NumPy broadcasts `base_prices` across both rows because the final dimension has three elements in each array.


## Interpretation

Broadcasting avoids manually duplicating values to match array shapes. This makes numerical code simpler and often more efficient. However, the dimensions must be compatible; otherwise, NumPy raises a shape-related error.

# 15. Array Operations in NumPy

## Definition

Array operations perform calculations directly on NumPy arrays. Common arithmetic operations include addition, subtraction, multiplication, division, and exponentiation.

For arrays with the same shape, these arithmetic operations generally work element by element.

## Formula

For two arrays \(A\) and \(B\):

\[
C = A + B
\]

\[
D = A \times B
\]

Here, each corresponding pair of elements is added or multiplied.

## Simple Numerical Example

Let `A = [2, 4, 6]` and `B = [1, 2, 3]`.

- Addition: `[3, 6, 9]`
- Subtraction: `[1, 2, 3]`
- Multiplication: `[2, 8, 18]`
- Division: `[2, 2, 2]`

## Business Example

A business can calculate total sales per product by adding arrays of sales from two different branches.

## AI/ML Example

Array operations are used to calculate prediction errors, transform features, update numerical parameters, and perform computations within Machine Learning algorithms.

```python
import numpy as np

a = np.array([2, 4, 6])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power:", a ** 2)
```

## Code Implementation Explanation

- `a + b` adds corresponding elements.
- `a - b` subtracts corresponding elements.
- `a * b` multiplies corresponding elements.
- `a / b` divides corresponding elements and normally produces floating-point results.
- `a ** 2` squares each element of `a`.


## Interpretation

NumPy performs element-wise arithmetic directly on arrays. These operations are useful when calculating totals, differences, ratios, and transformations for numerical datasets.

# 16. Universal Functions (ufuncs)

## Definition 

Universal functions, commonly called **ufuncs**, are NumPy functions designed to perform operations element by element on array values.

Examples include:

- `np.sqrt()` — calculates square roots.
- `np.exp()` — calculates exponential values.
- `np.log()` — calculates natural logarithms.
- `np.abs()` — calculates absolute values.
- `np.sin()` — calculates sine values.
- `np.maximum()` — compares corresponding elements and returns the larger value.

Ufuncs generally support array inputs and perform numerical operations efficiently.

## Formula

For the square-root function:

\[
y_i = \sqrt{x_i}
\]

For the absolute-value function:

\[
y_i = |x_i|
\]

## Simple Numerical Example

For the array `[4, 9, 16]`, applying `np.sqrt()` gives `[2, 3, 4]`.

For `[-2, 3, -5]`, applying `np.abs()` gives `[2, 3, 5]`.

## Business Example

A business analyst can use NumPy functions to transform numerical measurements, calculate magnitudes, and process large collections of values.

## AI/ML Example

Ufuncs are useful in feature transformations and mathematical calculations. For example, square roots and logarithms can transform skewed data, while exponential functions are used in probability calculations and some Machine Learning algorithms.

```python
import numpy as np

numbers = np.array([4, 9, 16, 25])
values = np.array([-2, 3, -5, 8])

print("Square Roots:", np.sqrt(numbers))
print("Absolute Values:", np.abs(values))
print("Exponential Values:", np.round(np.exp([0, 1, 2]), 2))
print("Natural Logarithms:", np.round(np.log([1, np.e, np.e**2]), 2))
```

## Code Implementation Explanation

- `np.sqrt(numbers)` calculates the square root of each element.
- `np.abs(values)` converts negative values into their non-negative magnitudes.
- `np.exp()` calculates the exponential function \(e^x\) for each element.
- `np.log()` calculates the natural logarithm of each positive element.
- `np.round(..., 2)` rounds the displayed numerical results to two decimal places.

## Interpretation

Ufuncs apply mathematical functions to array elements without requiring an explicit Python loop. They are useful for efficient numerical processing, data transformation, and AI/ML computations.
