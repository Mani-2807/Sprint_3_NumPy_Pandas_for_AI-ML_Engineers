## CSV files

### 1. Concept explanation
CSV stores tabular values as delimited text.

### 2. Real-world analogy
A simple text version of a spreadsheet.

### 3. Business example
Exchange sales reports with another team.

### 4. AI/ML example
Load labeled training datasets.

```python
df = pd.DataFrame({'id': [1, 2], 'amount': [10.5, 20.0]})
df.to_csv('orders.csv', index=False)
print(pd.read_csv('orders.csv'))
```

### 5. Output explanation
The saved data is read back into a DataFrame.

### 6. Best practices
Set `dtype` or `parse_dates` when appropriate.

### 7. Common mistakes
IDs such as '0012' can lose leading zeros if inferred as integers.

### 8. Advantages
Widely supported and readable.

### 9. Limitations
Data types and nested data are not preserved reliably.

## Excel files

### 1. Concept explanation
Excel files contain worksheets and may include formatting and formulas.

### 2. Real-world analogy
A workbook with multiple tabs.

### 3. Business example
Share a report with separate summary and detail sheets.

### 4. AI/ML example
Inspect business-provided datasets.

```python
df = pd.DataFrame({'store': ['A', 'B'], 'sales': [100, 140]})
df.to_excel('sales.xlsx', index=False, sheet_name='Sales')
print(pd.read_excel('sales.xlsx', sheet_name='Sales'))
```

### 5. Output explanation
The Sales worksheet is read back as a DataFrame.

### 6. Best practices
Install `openpyxl` for common `.xlsx` workflows.

### 7. Common mistakes
Reading a formula cell may return its stored value rather than recalculate formulas.

### 8. Advantages
Familiar business format.

### 9. Limitations
Can be slower and less compact than analytical formats.

## JSON, Parquet, and Pickle

### 1. Concept explanation
JSON is a text-based structured format; Parquet is a columnar analytical format; Pickle serializes Python objects.

### 2. Real-world analogy
JSON is a structured message, Parquet is an efficient data cabinet, and Pickle is a Python-specific package.

### 3. Business example
Exchange API data, store large analytics tables, or cache Python objects.

### 4. AI/ML example
Persist intermediate data in a pipeline.

```python
df = pd.DataFrame({'id': [1, 2], 'score': [0.8, 0.6]})
df.to_json('scores.json', orient='records')
print(pd.read_json('scores.json'))
try:
    df.to_parquet('scores.parquet', index=False)
    print(pd.read_parquet('scores.parquet'))
except ImportError:
    print("Parquet example skipped: install pyarrow or fastparquet to run it.")
df.to_pickle('scores.pkl')
print(pd.read_pickle('scores.pkl'))
```

### 5. Output explanation
Each read operation restores a DataFrame with the same two rows. Parquet needs an engine such as pyarrow or fastparquet.

### 6. Best practices
Prefer Parquet for large analytical tables; only load trusted Pickle files.

### 7. Common mistakes
Never unpickle untrusted files because loading them can execute code.

### 8. Advantages
Each format has a useful purpose.

### 9. Limitations
Pickle is Python-specific; Parquet requires an engine.

## SQL tables and multiple files

### 1. Concept explanation
Pandas can read SQL query results and combine data from multiple files.

### 2. Real-world analogy
Collect reports from several branches into one report.

### 3. Business example
Combine monthly exports or query orders from a database.

### 4. AI/ML example
Build a dataset from database tables and file drops.

```python
from pathlib import Path
folder = Path('monthly_reports')
folder.mkdir(exist_ok=True)
pd.DataFrame({'month': ['Jan'], 'sales': [100]}).to_csv(folder / 'jan.csv', index=False)
pd.DataFrame({'month': ['Feb'], 'sales': [120]}).to_csv(folder / 'feb.csv', index=False)
frames = [pd.read_csv(path) for path in sorted(folder.glob('*.csv'))]
combined = pd.concat(frames, ignore_index=True)
print(combined)
# SQL pattern (requires a database connection):
# df = pd.read_sql_query('SELECT * FROM orders', connection)
```

### 5. Output explanation
The combined DataFrame has the rows from both monthly CSV files. SQL example is a template and needs a live connection.

### 6. Best practices
Use `pathlib.Path` and parameterized SQL queries for user-provided values.

### 7. Common mistakes
Do not concatenate files with different column meanings without validating them.

### 8. Advantages
Automates repeatable data ingestion.

### 9. Limitations
Database access requires a connection and permissions.
