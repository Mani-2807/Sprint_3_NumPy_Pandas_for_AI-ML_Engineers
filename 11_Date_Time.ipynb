## Datetime, Timestamp, and date parsing

### 1. Concept explanation
Pandas timestamps represent points in time; `to_datetime` parses date-like strings into datetime values.

### 2. Real-world analogy
Convert written dates into a calendar system computers can sort.

### 3. Business example
Analyze order times consistently.

### 4. AI/ML example
Build time-based features for demand prediction.

```python
df = pd.DataFrame({'order_date': ['2026-01-05', '2026/01/08', '2026-02-02']})
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
print(df)
print(df['order_date'].dtype)
```

### 5. Output explanation
Date strings become datetime values; invalid inputs become `NaT` when coercion is used.

### 6. Best practices
Use explicit formats when inputs are known and consistent.

### 7. Common mistakes
Unparsed strings sort lexicographically rather than chronologically.

### 8. Advantages
Enables reliable time operations.

### 9. Limitations
Ambiguous dates such as 03/04/2026 need a clear convention.

## Timedelta

### 1. Concept explanation
A timedelta represents a duration or difference between times.

### 2. Real-world analogy
Measure how long a trip took.

### 3. Business example
Calculate shipping time and time-to-resolution.

### 4. AI/ML example
Use time gaps as model features or target durations.

```python
start = pd.Timestamp('2026-01-01 09:00')
end = pd.Timestamp('2026-01-03 12:30')
print(end - start)
print(start + pd.Timedelta(days=2))
```

### 5. Output explanation
The duration is 2 days 3 hours 30 minutes; adding two days gives January 3 at 09:00.

### 6. Best practices
Store durations with clear units.

### 7. Common mistakes
A duration is not the same as a calendar date.

### 8. Advantages
Simple duration arithmetic.

### 9. Limitations
Business calendars and daylight saving can complicate durations.

## Time zones

### 1. Concept explanation
Time-zone-aware timestamps represent local times with an offset/zone; conversion changes the displayed local time while preserving the instant.

### 2. Real-world analogy
The same meeting shown on clocks in two cities.

### 3. Business example
Compare transactions recorded across regions.

### 4. AI/ML example
Align event timestamps before time-series modeling.

```python
times = pd.Series(pd.to_datetime(['2026-01-01 12:00', '2026-01-01 18:00'], utc=True))
print(times)
print(times.dt.tz_convert('America/Chicago'))
```

### 5. Output explanation
UTC times are displayed and then converted to US Central time.

### 6. Best practices
Normalize cross-region events to UTC and convert for reporting.

### 7. Common mistakes
Naive and aware timestamps cannot always be compared directly.

### 8. Advantages
Consistent timeline analysis.

### 9. Limitations
Daylight-saving transitions need careful handling.

## Extracting date features

### 1. Concept explanation
Datetime accessors expose year, month, day, weekday, hour, and other calendar components.

### 2. Real-world analogy
Label a calendar entry by weekday or month.

### 3. Business example
Compare sales by weekday and season.

### 4. AI/ML example
Create calendar features for forecasting.

```python
df = pd.DataFrame({'timestamp': pd.to_datetime(['2026-01-05 09:30', '2026-01-10 15:45'])})
df['weekday'] = df['timestamp'].dt.day_name()
df['month'] = df['timestamp'].dt.month
df['hour'] = df['timestamp'].dt.hour
print(df)
```

### 5. Output explanation
The table includes weekday names, month numbers, and hour values.

### 6. Best practices
Consider cyclic encoding for hour or month in some models.

### 7. Common mistakes
Using future-derived information can cause leakage.

### 8. Advantages
Adds useful temporal signals.

### 9. Limitations
Calendar features alone may miss holidays and business-specific events.
