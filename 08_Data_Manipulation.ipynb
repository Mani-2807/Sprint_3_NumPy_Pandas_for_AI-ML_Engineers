## Sample dataset

```python
df = pd.DataFrame({
    'Customer Name': ['Ava', 'Ben', 'Ben', 'Dia'],
    'City': ['Dallas', 'Austin', 'Austin', 'Houston'],
    'Spend': [120, 250, 250, 175],
    'Status': ['active', 'active', 'active', 'inactive']
})
print(df)
```

## Rename, drop rows, and drop columns

### 1. Concept explanation
`rename` changes labels; `drop` removes selected rows or columns.

### 2. Real-world analogy
Edit a report heading or remove a duplicate page.

### 3. Business example
Remove irrelevant fields or invalid records.

### 4. AI/ML example
Keep only relevant model features after review.

```python
renamed = df.rename(columns={'Customer Name': 'customer', 'Spend': 'spend'})
without_city = renamed.drop(columns=['City'])
without_row = renamed.drop(index=0)
print(renamed.columns.tolist())
print(without_city)
print(without_row)
```

### 5. Output explanation
The column names change; separate outputs omit the selected column or row.

### 6. Best practices
Assign the result or use `inplace` only when mutation is intended.

### 7. Common mistakes
Dropping rows by index label is not the same as dropping by position.

### 8. Advantages
Straightforward dataset cleanup.

### 9. Limitations
Dropping data can remove useful information.

## Insert columns and assign()

### 1. Concept explanation
Add a column directly or use `assign` to produce a DataFrame with new calculated columns.

### 2. Real-world analogy
Add a total column to a form.

### 3. Business example
Calculate order totals or flags.

### 4. AI/ML example
Engineer new features from existing data.

```python
df2 = df.assign(spend_with_tax=df['Spend'] * 1.08, is_active=df['Status'].eq('active'))
df2.insert(1, 'record_id', [101, 102, 103, 104])
print(df2)
```

### 5. Output explanation
The table contains calculated spend_with_tax, is_active, and record_id columns.

### 6. Best practices
Use descriptive feature names and document units.

### 7. Common mistakes
Repeatedly inserting many columns can fragment memory in some workflows.

### 8. Advantages
Readable feature engineering.

### 9. Limitations
New features can introduce leakage if they use future information.

## apply() and map()

### 1. Concept explanation
`apply` applies a function along an axis or to values; Series `map` maps each value using a function or dictionary.

### 2. Real-world analogy
Apply a rule to each record or translate labels using a lookup chart.

### 3. Business example
Classify spending levels or standardize categories.

### 4. AI/ML example
Create features from raw values.

```python
df3 = df.assign(spend_band=df['Spend'].apply(lambda x: 'high' if x >= 200 else 'standard'))
city_code = df['City'].map({'Dallas': 'DAL', 'Austin': 'AUS', 'Houston': 'HOU'})
print(df3[['Customer Name', 'Spend', 'spend_band']])
print(city_code.tolist())
```

### 5. Output explanation
Spending values become high or standard; cities map to short codes.

### 6. Best practices
Use vectorized operations when they express the same logic clearly.

### 7. Common mistakes
A missing dictionary key maps to NaN by default.

### 8. Advantages
Flexible transformations.

### 9. Limitations
Python functions through apply can be slower than vectorized methods.

## replace() and duplicate handling

### 1. Concept explanation
`replace` substitutes values; `duplicated` detects duplicates and `drop_duplicates` removes them.

### 2. Real-world analogy
Correct a mislabeled entry and remove repeated forms.

### 3. Business example
Standardize status labels and deduplicate imported records.

### 4. AI/ML example
Prevent duplicated examples from biasing training.

```python
clean = df.replace({'Status': {'inactive': 'not_active'}})
print('Duplicate rows:', df.duplicated().sum())
print(df.drop_duplicates())
```

### 5. Output explanation
The status label is changed in the copy; duplicate full rows are reduced.

### 6. Best practices
Define which columns make a record unique before deduplicating.

### 7. Common mistakes
Two rows with the same values may still represent legitimate separate events.

### 8. Advantages
Improves consistency.

### 9. Limitations
Over-aggressive deduplication can delete valid transactions.
