# 1. What Is Pandas?

## Definition and Simple Explanation

Pandas is a Python library used to store, organize, analyze, and manipulate structured data.

It provides two main data structures:

- **Series:** A one-dimensional labeled collection of values.
- **DataFrame:** A two-dimensional table containing rows and columns.

Pandas is built on top of NumPy and works well with tabular data.

## Formula

Not applicable. Pandas is a data analysis library rather than a mathematical formula.

## Simple Numerical Example

Consider the marks of three students:

- Student A: 80
- Student B: 90
- Student C: 75

Pandas can store these marks in a table and help calculate the average.

## Business Example

A company can use Pandas to organize sales records, calculate total revenue, and identify its best-selling products.

## AI/ML Example

An AI/ML engineer can use Pandas to load a dataset, inspect its features, identify missing values, and prepare the data before training a model.

````python
import pandas as pd

student_data = {
    "Student": ["A", "B", "C"],
    "Marks": [80, 90, 75]
}

df = pd.DataFrame(student_data)

print(df)
print("Average marks:", df["Marks"].mean())
````

### Code Implementation Explanation

- `student_data` is a Python dictionary containing student names and marks.
- `pd.DataFrame(student_data)` converts the dictionary into a Pandas DataFrame.
- `df["Marks"]` selects the Marks column.
- `.mean()` calculates the average of the marks.

### Interpretation

Pandas makes it easy to store information in a table and perform calculations on selected columns. Here, we organize student records and calculate the average marks without manually adding each value.

# 2. Why Do We Use Pandas?

## Definition and Simple Explanation

Pandas simplifies data analysis by providing functions to organize, filter, summarize, and transform data.

Important capabilities include:

- Handling rows and columns.
- Selecting specific data.
- Calculating summary statistics.
- Detecting and handling missing values.
- Combining datasets.
- Reading and writing files.
- Preparing data for machine learning.

## Formula

Not applicable. Pandas provides tools and operations rather than one specific formula.

## Simple Numerical Example

Suppose daily sales are:

100, 150, 200

Pandas can calculate the total sales:

Total sales = 100 + 150 + 200 = 450

## Business Example

A store can analyze daily sales to find total revenue, average sales, and the day with the highest sales.

## AI/ML Example

Before training a model to predict house prices, an engineer can use Pandas to organize house features such as area, number of bedrooms, and price.

````python
import pandas as pd

sales = pd.DataFrame({
    "Day": ["Monday", "Tuesday", "Wednesday"],
    "Sales": [100, 150, 200]
})

total_sales = sales["Sales"].sum()
average_sales = sales["Sales"].mean()
highest_sales = sales["Sales"].max()

print("Total sales:", total_sales)
print("Average sales:", average_sales)
print("Highest daily sales:", highest_sales)
````

### Code Implementation Explanation

- `pd.DataFrame()` creates a table of daily sales.
- `.sum()` calculates the total sales.
- `.mean()` calculates the average sales.
- `.max()` finds the highest sales value.

### Interpretation

Pandas can summarize business data with a few simple commands. These same operations are useful for exploring datasets before applying statistical methods or building machine learning models.

# 3. Pandas Series

## Definition and Simple Explanation

A **Series** is a one-dimensional labeled data structure in Pandas.

It contains:
- Values
- An index that identifies each value

A Series can store integers, decimal numbers, text, or other data types.

## Formula

Not applicable. A Series is a data structure.

## Simple Numerical Example

Consider the marks:

80, 90, 75

A Series can store these values with index labels 0, 1, and 2.

## Business Example

A business can represent the monthly revenue of a store as a Series, where each value corresponds to a month.

## AI/ML Example

A Series can represent one feature of a dataset, such as house prices, student marks, or customer ages.

````python
import pandas as pd

marks = pd.Series([80, 90, 75])

print(marks)

print("First mark:", marks.iloc[0])
print("Number of values:", marks.size)
print("Average marks:", marks.mean())
````

### Code Implementation Explanation

- `pd.Series([80, 90, 75])` creates a Series containing three values.
- `.iloc[0]` selects the first value by its position.
- `.size` returns the number of elements.
- `.mean()` calculates the average of the values.

### Interpretation

A Series is useful when working with a single sequence of labeled values. Its index allows us to identify and retrieve individual elements.

## Creating a Series with Custom Index Labels

A Series can use custom labels instead of the default numerical index.

For example, student names can be used as index labels for their marks.

````python
import pandas as pd

marks = pd.Series(
    [80, 90, 75],
    index=["Anu", "Ravi", "Maya"]
)

print(marks)
print("Ravi's marks:", marks.loc["Ravi"])
````

### Code Implementation Explanation

- `index=["Anu", "Ravi", "Maya"]` assigns a label to each value.
- `.loc["Ravi"]` retrieves the value associated with the label `"Ravi"`.

### Interpretation

Custom index labels make data easier to understand. Instead of remembering a student's numerical position, we can retrieve their marks using their name.

# 4. Pandas DataFrame

## Definition and Simple Explanation

A **DataFrame** is a two-dimensional tabular data structure with rows and columns.

It resembles a spreadsheet or database table.

Each column can contain a different data type. For example:

- Name: text
- Age: integer
- Marks: numerical value

## Formula

Not applicable. A DataFrame is a data structure.

## Simple Numerical Example

| Student | Age | Marks |
|---|---:|---:|
| Anu | 20 | 80 |
| Ravi | 21 | 90 |
| Maya | 19 | 75 |

This table can be represented as a DataFrame with three rows and three columns.

## Business Example

A sales DataFrame can contain product names, quantities sold, unit prices, and revenue.

## AI/ML Example

A machine learning dataset is often represented as a DataFrame. Rows usually represent observations, while columns represent features or a target variable.

````python
import pandas as pd

students = pd.DataFrame({
    "Name": ["Anu", "Ravi", "Maya"],
    "Age": [20, 21, 19],
    "Marks": [80, 90, 75]
})

print(students)
print("Number of rows and columns:", students.shape)
````

### Code Implementation Explanation

- `pd.DataFrame({...})` converts the dictionary into a table.
- Each dictionary key becomes a column name.
- Each list provides the values for its corresponding column.
- `.shape` returns a tuple containing the number of rows and columns.

### Interpretation

A DataFrame stores related information in a structured table. In this example, each row represents one student, and each column represents a student attribute.

# 5. Different Ways to Create DataFrames

## Definition and Simple Explanation

Pandas supports several ways to create a DataFrame, including:

1. From a dictionary of lists.
2. From a list of dictionaries.
3. From a list of lists with column names.
4. From an existing NumPy array.

The best method depends on how the data is already organized.

## Formula

Not applicable. Creating a DataFrame is a data preparation operation.

## Simple Numerical Example

Consider two products:

- Pen: quantity 10
- Notebook: quantity 5

These records can be converted into a DataFrame using a dictionary or a list of dictionaries.

## Business Example

A company can convert product inventory records into a DataFrame for stock analysis.

## AI/ML Example

An engineer can convert collected observations into a DataFrame before cleaning and preparing the data for a model.

````python
import pandas as pd
import numpy as np

# Method 1: Dictionary of lists
df1 = pd.DataFrame({
    "Product": ["Pen", "Notebook"],
    "Quantity": [10, 5]
})

# Method 2: List of dictionaries
df2 = pd.DataFrame([
    {"Product": "Pen", "Quantity": 10},
    {"Product": "Notebook", "Quantity": 5}
])

# Method 3: List of lists
df3 = pd.DataFrame(
    [["Pen", 10], ["Notebook", 5]],
    columns=["Product", "Quantity"]
)

# Method 4: NumPy array
array = np.array([
    [10, 100],
    [5, 200]
])

df4 = pd.DataFrame(
    array,
    columns=["Quantity", "Price"]
)

print("DataFrame from dictionary:")
print(df1)

print("\nDataFrame from list of dictionaries:")
print(df2)

print("\nDataFrame from list of lists:")
print(df3)

print("\nDataFrame from NumPy array:")
print(df4)
````

### Code Implementation Explanation

- `df1` is created from a dictionary of lists.
- `df2` is created from a list of dictionaries, where each dictionary represents one row.
- `df3` uses a list of lists and explicitly specifies the column names.
- `df4` converts a NumPy array into a DataFrame with named columns.

### Interpretation

Pandas can work with data from several Python structures. This flexibility is useful because real-world data may arrive as dictionaries, lists, NumPy arrays, or files.

# 6. Data Types in Pandas

## Definition and Simple Explanation

A data type describes the kind of value stored in a column or Series.

Common Pandas data types include:

- `int64`: Whole numbers
- `float64`: Decimal numbers
- `object`: Often used for text or mixed Python objects
- `bool`: True or False values
- `datetime64`: Date and time values
- `category`: Values belonging to a limited set of categories
- `string`: Text values using Pandas' string data type

The exact dtype may vary depending on the data and the Pandas version.

## Formula

Not applicable. Data types describe how values are represented.

## Simple Numerical Example

A table may contain:
- Age: 21 — integer
- Price: 99.50 — decimal
- Name: "Ravi" — text
- Passed: True — Boolean

## Business Example

A sales dataset needs numerical types for prices and quantities, text for product names, and appropriate date types for transaction dates.

## AI/ML Example

Correct data types help prevent data preparation errors. For example, numerical features must be represented appropriately before mathematical operations or model training.

````python
import pandas as pd

data = pd.DataFrame({
    "Age": [20, 21, 19],
    "Price": [99.50, 150.75, 75.00],
    "Name": ["Anu", "Ravi", "Maya"],
    "Passed": [True, True, False]
})

print("Data types:")
print(data.dtypes)

# Convert Age to a decimal type
data["Age"] = data["Age"].astype("float64")

print("\nData types after conversion:")
print(data.dtypes)
````

### Code Implementation Explanation

- `data.dtypes` displays the data type of each column.
- `.astype("float64")` converts the Age column to a floating-point data type.
- The converted values can now be represented as decimal numbers.

### Interpretation

Checking data types helps us understand how Pandas interprets each column. Use type conversion when a column needs a different representation, but confirm that the values can be converted safely.

# 7. Index in Pandas

## Definition and Simple Explanation

An index is a set of labels used to identify rows in a Series or DataFrame.

By default, Pandas usually creates a numerical index starting from 0.

An index can also contain custom labels, such as student IDs or product codes.

## Formula

Not applicable. An index identifies rows rather than calculating a mathematical result.

## Simple Numerical Example

| Index | Student | Marks |
|---:|---|---:|
| 0 | Anu | 80 |
| 1 | Ravi | 90 |
| 2 | Maya | 75 |

Here, 0, 1, and 2 are the default row labels.

## Business Example

A company can use an order ID as an index to identify individual orders.

## AI/ML Example

Indexes help maintain row labels while exploring or cleaning datasets. However, the index is not automatically a model feature; it should be included as a feature only when appropriate.

````python
import pandas as pd

students = pd.DataFrame({
    "Student": ["Anu", "Ravi", "Maya"],
    "Marks": [80, 90, 75]
})

print("Default index:")
print(students.index)

students.index = ["S101", "S102", "S103"]

print("\nDataFrame with custom index:")
print(students)

print("\nRecord for S102:")
print(students.loc["S102"])
````

### Code Implementation Explanation

- `.index` displays the row labels.
- `students.index = [...]` replaces the default index with custom labels.
- `.loc["S102"]` retrieves the row whose index label is `"S102"`.

### Interpretation

Custom indexes can make records easier to identify. Remember that `.loc` selects by label, while `.iloc` selects by numerical position.

# 8. Columns in Pandas

## Definition and Simple Explanation

Columns represent individual variables or attributes in a DataFrame.

For example, a student DataFrame may contain:

- Name
- Age
- Marks

Each column contains values for one attribute across the rows.

## Formula

Not applicable. Columns organize attributes rather than calculate a mathematical result.

## Simple Numerical Example

If a DataFrame has the columns `Product`, `Quantity`, and `Price`, the Quantity column may contain 10, 5, and 8.

Selecting that column allows us to analyze quantities separately.

## Business Example

A sales table may contain Product, Quantity, Price, and Revenue columns. Selecting Revenue allows a business to analyze income without selecting unrelated columns.

## AI/ML Example

DataFrame columns often represent model features, such as age, income, and purchase frequency. The target column represents the value the model is expected to predict.

````python
import pandas as pd

students = pd.DataFrame({
    "Name": ["Anu", "Ravi", "Maya"],
    "Age": [20, 21, 19],
    "Marks": [80, 90, 75]
})

# Display all column names
print("Column names:", students.columns.tolist())

# Select one column
print("\nMarks column:")
print(students["Marks"])

# Select multiple columns
print("\nName and Marks:")
print(students[["Name", "Marks"]])

# Rename a column
students = students.rename(columns={"Marks": "Exam_Marks"})

print("\nAfter renaming:")
print(students)
````

### Code Implementation Explanation

- `.columns` returns the DataFrame's column labels.
- `.tolist()` converts the column labels into a Python list.
- `students["Marks"]` selects one column as a Series.
- `students[["Name", "Marks"]]` selects multiple columns as a DataFrame.
- `.rename(columns={...})` changes a column name.

### Interpretation

Selecting specific columns helps us focus on relevant information. Renaming columns can also make datasets more consistent and easier to understand.

# 9. Reading Data with Pandas

## Definition and Simple Explanation

Reading data means loading information from an external source into a Pandas object.

Pandas supports many data formats, including:

- CSV: `pd.read_csv()`
- Excel: `pd.read_excel()`
- JSON: `pd.read_json()`
- SQL tables: `pd.read_sql()`
- Parquet: `pd.read_parquet()`

The available format and its dependencies determine which method should be used.

## Formula

Not applicable. Reading data is a data input operation.

## Simple Numerical Example

A CSV file may contain:

```text
Product,Quantity
Pen,10
Notebook,5
```

Pandas can load these records into a DataFrame.

## Business Example

A company can load sales data exported from its billing system and analyze the total quantity sold for each product.

## AI/ML Example

An engineer can load a training dataset from a CSV file before checking data types, missing values, and feature distributions.

````python
import pandas as pd
from io import StringIO

# Example CSV content stored in memory
csv_data = """Product,Quantity,Price
Pen,10,5
Notebook,5,40
Pencil,8,3
"""

# Read CSV data
sales = pd.read_csv(StringIO(csv_data))

print("CSV data:")
print(sales)

# Example JSON content stored in memory
json_data = """
[
    {"Product": "Pen", "Quantity": 10},
    {"Product": "Notebook", "Quantity": 5}
]
"""

# Read JSON data
products = pd.read_json(StringIO(json_data))

print("\nJSON data:")
print(products)
````

### Code Implementation Explanation

- `StringIO` lets Pandas read a string as if it were a file. This example does not require an existing CSV or JSON file.
- `pd.read_csv()` reads CSV-formatted data into a DataFrame.
- `pd.read_json()` reads JSON-formatted data into a Pandas object.
- The variable `sales` stores the CSV records.
- The variable `products` stores the JSON records.

### Reading a Real CSV File

For a CSV file located in your working directory, use:

```python
sales = pd.read_csv("sales.csv")
```

For an Excel file, use:

```python
sales = pd.read_excel("sales.xlsx")
```

Excel support commonly requires the `openpyxl` package for `.xlsx` files. If it is missing, install it in your environment.

### Interpretation

Pandas provides convenient functions for loading data from different formats. In real projects, the file path, column names, encoding, and file format should be checked if loading fails.

# 10. Writing Data with Pandas

## Definition and Simple Explanation

Writing data means saving a Series or DataFrame to a file or another storage system.

Common methods include:

- `.to_csv()`: Save as CSV.
- `.to_excel()`: Save as Excel.
- `.to_json()`: Save as JSON.
- `.to_parquet()`: Save as Parquet.

Writing data is useful for storing cleaned datasets, sharing reports, and saving intermediate results.

## Formula

Not applicable. Writing data is a data output operation.

## Simple Numerical Example

A DataFrame containing product quantities can be saved as a CSV file so it can be opened later in a spreadsheet application.

## Business Example

An analyst can save a daily sales summary as a CSV file and share it with a business team.

## AI/ML Example

An AI/ML engineer can save a cleaned dataset for reuse in a later model-training step.

````python
import pandas as pd

sales = pd.DataFrame({
    "Product": ["Pen", "Notebook", "Pencil"],
    "Quantity": [10, 5, 8],
    "Price": [5, 40, 3]
})

# Save as CSV without writing the DataFrame index
sales.to_csv("sales_output.csv", index=False)

# Save as JSON
sales.to_json("sales_output.json", orient="records", indent=4)

print("CSV and JSON files saved.")
````

### Code Implementation Explanation

- `sales.to_csv("sales_output.csv", index=False)` saves the DataFrame as a CSV file.
- `index=False` prevents Pandas from adding the row index as an extra column in the saved file.
- `sales.to_json()` saves the data in JSON format.
- `orient="records"` represents each row as a JSON object.
- `indent=4` formats the JSON file with indentation for readability.

### Saving an Excel File

To save the same DataFrame as Excel, use:

```python
sales.to_excel("sales_output.xlsx", index=False)
```

This commonly requires the `openpyxl` package for `.xlsx` files.

### Reading a Saved File Again

You can load the CSV file using:

```python
loaded_sales = pd.read_csv("sales_output.csv")
print(loaded_sales)
```

### Interpretation

Writing data allows us to preserve our work and reuse it later. The saved files are created in the current working directory unless another path is specified. In a notebook environment, file persistence and download options depend on the platform.
